```python
import pandas as pd
import re
import os

# Load the datasets
raw_folder = '/content/drive/MyDrive/corporate_risk_graphrag/data/raw'
reg_df = pd.read_csv(os.path.join(raw_folder, 'registered company.csv'), low_memory=False)
fin_df = pd.read_csv(os.path.join(raw_folder, 'financial_data.csv'))

# Define the cleaning function
def clean_company_name(name):
    if pd.isna(name):
        return ""

    # 1. Convert to uppercase
    name = str(name).upper()

    # 2. Remove punctuation & special characters (keep alphanumeric and spaces)
    name = re.sub(r'[^\w\s]', ' ', name)

    # 3. Remove common corporate suffixes using word boundaries
    # This prevents removing "LTD" from the middle of a name accidentally
    suffixes = ['LIMITED', 'LTD', 'PRIVATE', 'PVT', 'CORPORATION', 'CORP', 'COMPANY', 'CO', 'LLP', 'LLC']
    pattern = r'\b(?:' + '|'.join(suffixes) + r')\b'
    name = re.sub(pattern, '', name)

    # 4. Remove extra spaces
    name = ' '.join(name.split())

    return name

# Apply the cleaning function to both datasets
print("Cleaning Registration Data names...")
reg_df['CLEAN_NAME'] = reg_df['COMPANY_NAME'].apply(clean_company_name)

print("Cleaning Financial Data names...")
fin_df['CLEAN_NAME'] = fin_df['Company Name'].apply(clean_company_name)

# Let's look at the before and after to see the magic!
print("\n--- Registration Data: Before & After ---")
display(reg_df[['COMPANY_NAME', 'CLEAN_NAME']].head(5))

print("\n--- Financial Data: Before & After ---")
display(fin_df[['Company Name', 'CLEAN_NAME']].head(10))
```

```python
# Install RapidFuzz
!pip install rapidfuzz -q

import pandas as pd
from rapidfuzz import process, fuzz
import os

# Load the cleaned data (re-running the cleaning step to ensure we have the data in memory)
raw_folder = '/content/drive/MyDrive/corporate_risk_graphrag/data/raw'
reg_df = pd.read_csv(os.path.join(raw_folder, 'registered company.csv'), low_memory=False)
fin_df = pd.read_csv(os.path.join(raw_folder, 'financial_data.csv'))

# Quick clean function (from previous step)
import re
def clean_company_name(name):
    if pd.isna(name): return ""
    name = str(name).upper()
    name = re.sub(r'[^\w\s]', ' ', name)
    suffixes = ['LIMITED', 'LTD', 'PRIVATE', 'PVT', 'CORPORATION', 'CORP', 'COMPANY', 'CO', 'LLP', 'LLC']
    pattern = r'\b(?:' + '|'.join(suffixes) + r')\b'
    name = re.sub(pattern, '', name)
    return ' '.join(name.split())

reg_df['CLEAN_NAME'] = reg_df['COMPANY_NAME'].apply(clean_company_name)
fin_df['CLEAN_NAME'] = fin_df['Company Name'].apply(clean_company_name)

# Drop any rows in registration data with empty clean names to speed up search
reg_df_clean = reg_df[reg_df['CLEAN_NAME'] != ""].copy()

print("Starting Entity Resolution...")

# Lists to store results
matched_data = []

# Create a dictionary of clean names to original names and CINs for fast lookup
reg_names_list = reg_df_clean['CLEAN_NAME'].tolist()

# Loop through our 50 financial companies
for idx, fin_row in fin_df.iterrows():
    fin_clean = fin_row['CLEAN_NAME']
    fin_original = fin_row['Company Name']

    # STAGE 3: EXACT MATCH
    # Check if the exact clean name exists in the registration list
    exact_match = reg_df_clean[reg_df_clean['CLEAN_NAME'] == fin_clean]

    if not exact_match.empty:
        # Match found!
        cin = exact_match.iloc[0]['CORPORATE_IDENTIFICATION_NUMBER']
        reg_original = exact_match.iloc[0]['COMPANY_NAME']
        match_type = "Exact"
        score = 100.0
    else:
        # STAGE 4: FUZZY MATCH
        # If no exact match, use RapidFuzz to find the closest name
        # We use process.extractOne to get the best single match
        best_match = process.extractOne(fin_clean, reg_names_list, scorer=fuzz.WRatio)

        if best_match:
            best_name, score, best_idx = best_match
            # Only accept if score is >= 85 (threshold to prevent bad matches)
            if score >= 85:
                matched_reg_row = reg_df_clean.iloc[best_idx]
                cin = matched_reg_row['CORPORATE_IDENTIFICATION_NUMBER']
                reg_original = matched_reg_row['COMPANY_NAME']
                match_type = "Fuzzy"
            else:
                cin = "NO MATCH"
                reg_original = "NO MATCH"
                match_type = "Failed"
                score = score
        else:
            cin = "NO MATCH"
            reg_original = "NO MATCH"
            match_type = "Failed"
            score = 0.0

    # Append to our results list
    matched_data.append({
        'Financial_Company_Name': fin_original,
        'Registration_Company_Name': reg_original,
        'CIN': cin,
        'Match_Type': match_type,
        'Match_Score': score
    })

# Create a DataFrame from the results
match_results_df = pd.DataFrame(matched_data)

print("\n--- ENTITY RESOLUTION COMPLETE ---")
display(match_results_df)
```

```python
import pandas as pd

print("Fusing matched datasets...")

# 1. Merge match results back into the financial dataframe
fused_df = fin_df.merge(match_results_df, left_on='Company Name', right_on='Financial_Company_Name')

# 2. Keep only the columns we need from financial data + match metadata (FIXED: Added CIN and Registration_Company_Name)
fin_cols_to_keep = [
    'Financial_Company_Name', 'Ticker', 'Sector', 'Industry',
    'Total Revenue', 'Net Profit', 'Total Assets', 'Total Debt',
    'Debt to Equity', 'Current Ratio', 'Return on Equity', 'Profit Margins',
    'CIN', 'Registration_Company_Name', 'Match_Type', 'Match_Score'
]
fused_df = fused_df[fin_cols_to_keep]

# 3. Filter the massive registration dataframe to only include our matched CINs
matched_cins = fused_df['CIN'][fused_df['CIN'] != 'NO MATCH'].unique()
reg_filtered = reg_df[reg_df['CORPORATE_IDENTIFICATION_NUMBER'].isin(matched_cins)].copy()

# 4. Select the columns we need from registration data
reg_cols_to_keep = [
    'CORPORATE_IDENTIFICATION_NUMBER',
    'COMPANY_STATUS', 'DATE_OF_REGISTRATION', 'REGISTERED_STATE',
    'AUTHORIZED_CAP', 'PAIDUP_CAPITAL', 'PRINCIPAL_BUSINESS_ACTIVITY_AS_PER_CIN'
]
reg_filtered = reg_filtered[reg_cols_to_keep]

# 5. Merge Financial + Registration data on CIN
# We use a left merge so we keep all 50 companies, even if they failed to match
master_df = fused_df.merge(reg_filtered, left_on='CIN', right_on='CORPORATE_IDENTIFICATION_NUMBER', how='left')

# Reorder columns for better readability
final_columns = [
    'Financial_Company_Name', 'Registration_Company_Name', 'CIN',
    'Ticker', 'Sector', 'Industry', 'REGISTERED_STATE', 'COMPANY_STATUS', 'DATE_OF_REGISTRATION',
    'PRINCIPAL_BUSINESS_ACTIVITY_AS_PER_CIN', 'AUTHORIZED_CAP', 'PAIDUP_CAPITAL',
    'Total Revenue', 'Net Profit', 'Total Assets', 'Total Debt',
    'Debt to Equity', 'Current Ratio', 'Return on Equity', 'Profit Margins',
    'Match_Type', 'Match_Score'
]
# Ensure all columns exist before reordering
final_columns = [col for col in final_columns if col in master_df.columns]
master_df = master_df[final_columns]

print("\n--- MASTER CORPORATE DATASET CREATED ---")
print(f"Total unified companies: {len(master_df)}")
display(master_df.head(10))

# 6. Save the Master Dataset to Google Drive
save_path = '/content/drive/MyDrive/corporate_risk_graphrag/data/processed/master_corporate_dataset.csv'
master_df.to_csv(save_path, index=False)
print(f"\n✓ Master dataset saved to: {save_path}")
```

```python
# Install the neo4j driver if not already installed
!pip install neo4j

import pandas as pd
from neo4j import GraphDatabase

# 1. Load the Master Dataset
print("Loading Master Dataset...")
master_path = '/content/drive/MyDrive/corporate_risk_graphrag/data/processed/master_corporate_dataset.csv'
master_df = pd.read_csv(master_path)
master_df = master_df.where(pd.notnull(master_df), None)

# 2. Neo4j Connection Details
NEO4J_URI = "neo4j+s://240b1b1e.databases.neo4j.io"
NEO4J_USERNAME = "240b1b1e"
NEO4J_PASSWORD = "T0-hXHZMe0Oib5frLfcxD9vrFNTywqVOr_hp4ExFu74"
NEO4J_DATABASE = "240b1b1e"

# 3. Connect to Neo4j
print("Connecting to Neo4j AuraDB...")
driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USERNAME, NEO4J_PASSWORD))

try:
    driver.verify_connectivity()
    print("✓ Successfully connected to Neo4j!")
except Exception as e:
    print(f"Connection failed: {e}")

# 4. Define the function to create the Knowledge Graph (FIXED $row SYNTAX)
def create_knowledge_graph(tx, row):
    query = """
    MERGE (c:Company {cin: $row.CIN})
    SET c.name = $row.Registration_Company_Name,
        c.ticker = $row.Ticker,
        c.status = $row.COMPANY_STATUS,
        c.match_score = $row.Match_Score

    FOREACH (_ IN CASE WHEN $row.Industry IS NOT NULL THEN [1] ELSE [] END |
        MERGE (i:Industry {name: $row.Industry})
        MERGE (c)-[:OPERATES_IN]->(i)
    )

    FOREACH (_ IN CASE WHEN $row.REGISTERED_STATE IS NOT NULL THEN [1] ELSE [] END |
        MERGE (s:State {name: $row.REGISTERED_STATE})
        MERGE (c)-[:REGISTERED_IN]->(s)
    )

    FOREACH (_ IN CASE WHEN $row.CIN IS NOT NULL THEN [1] ELSE [] END |
        MERGE (f:FinancialRecord {company_cin: $row.CIN, period: "Latest"})
        SET f.revenue = $row.`Total Revenue`,
            f.net_profit = $row.`Net Profit`,
            f.total_debt = $row.`Total Debt`,
            f.debt_to_equity = $row.`Debt to Equity`,
            f.profit_margins = $row.`Profit Margins`
        MERGE (c)-[:HAS_FINANCIAL_RECORD]->(f)
    )
    """
    tx.run(query, parameters={"row": row.to_dict()})

# 5. Execute the graph creation
print("\nBuilding Knowledge Graph in Neo4j...")
with driver.session(database=NEO4J_DATABASE) as session:
    for index, row in master_df.iterrows():
        if row['CIN'] and row['CIN'] != "NO MATCH":
            session.execute_write(create_knowledge_graph, row)

driver.close()
print("\n✓ Knowledge Graph construction complete!")
```

```python
# Install LangGraph
!pip install langgraph -q

import os
import warnings
from langchain_groq import ChatGroq
from langchain_community.graphs import Neo4jGraph
from langchain_community.chains.graph_qa.cypher import GraphCypherQAChain
from langchain_core.prompts.prompt import PromptTemplate
from langgraph.graph import StateGraph, END
from typing import TypedDict, Literal

warnings.filterwarnings("ignore")

# 1. Credentials
os.environ["GROQ_API_KEY"] = "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"
NEO4J_URI = "neo4j+s://240b1b1e.databases.neo4j.io"
NEO4J_USERNAME = "240b1b1e"
NEO4J_PASSWORD = "T0-hXHZMe0Oib5frLfcxD9vrFNTywqVOr_hp4ExFu74"
NEO4J_DATABASE = "240b1b1e"

# 2. Connect to Neo4j and LLM
graph = Neo4jGraph(url=NEO4J_URI, username=NEO4J_USERNAME, password=NEO4J_PASSWORD, database=NEO4J_DATABASE)
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0)

# 3. Custom Cypher Prompt (from our previous step)
CYPHER_GENERATION_TEMPLATE = """
You are an expert Neo4j Cypher developer.
Task: Generate Cypher code to query a Neo4j Knowledge Graph to answer the user's question.

Graph Schema:
Node labels: Company, Industry, State, FinancialRecord
Relationships:
- (Company)-[:OPERATES_IN]->(Industry)
- (Company)-[:REGISTERED_IN]->(State)
- (Company)-[:HAS_FINANCIAL_RECORD]->(FinancialRecord)

Company properties: name, cin, ticker, status, match_score
FinancialRecord properties: revenue, net_profit, total_debt, debt_to_equity, profit_margins, period

CRITICAL INSTRUCTIONS FOR COMPANY NAMES:
The user may type partial company names. ALWAYS use `TOLOWER(c.name) CONTAINS TOLOWER('user_input')` in the WHERE clause. NEVER use exact matches.

User Question: {question}

Cypher Query (Return ONLY the Cypher code, no explanations, no backticks):
"""
cypher_prompt = PromptTemplate(template=CYPHER_GENERATION_TEMPLATE, input_variables=["question"])

# 4. Create the GraphRAG Tool (This will be used by our agents)
graph_qa = GraphCypherQAChain.from_llm(
    llm=llm,
    graph=graph,
    verbose=False,
    allow_dangerous_requests=True,
    cypher_prompt=cypher_prompt
)

# 5. Define the LangGraph State
class AgentState(TypedDict):
    query: str
    agent_type: str
    response: str

# 6. Define the Planner Agent (The Router)
def planner_agent(state: AgentState):
    print("--- PLANNER AGENT ANALYZING QUERY ---")
    query = state['query']

    # Use LLM to classify the query
    routing_prompt = f"""
    Analyze the following user query and classify it into one of three categories:
    1. "COMPANY" - If the question is about registration, location, state, industry, or company status.
    2. "FINANCIAL" - If the question is about revenue, debt, profit, margins, or financial ratios.
    3. "RISK" - If the question asks to assess, evaluate, or explain the risk of a company.

    Query: "{query}"

    Respond with ONLY ONE WORD: COMPANY, FINANCIAL, or RISK.
    """
    response = llm.invoke(routing_prompt)
    agent_type = response.content.strip().upper()

    if agent_type not in ["COMPANY", "FINANCIAL", "RISK"]:
        agent_type = "COMPANY" # Default

    print(f"-> Decision: Route to {agent_type} Agent")
    return {"agent_type": agent_type}

# 7. Define the Worker Agents
def company_agent(state: AgentState):
    print("--- COMPANY AGENT EXECUTING ---")
    response = graph_qa.invoke({"query": state['query']})
    return {"response": response['result']}

def financial_agent(state: AgentState):
    print("--- FINANCIAL AGENT EXECUTING ---")
    response = graph_qa.invoke({"query": state['query']})
    return {"response": response['result']}

def risk_agent(state: AgentState):
    print("--- RISK AGENT EXECUTING ---")
    # For now, the Risk Agent queries the graph for financials to reason about risk
    response = graph_qa.invoke({"query": state['query']})
    return {"response": response['result']}

# 8. Define the Routing Function
def route_query(state: AgentState) -> Literal["company_agent", "financial_agent", "risk_agent"]:
    if state['agent_type'] == "COMPANY":
        return "company_agent"
    elif state['agent_type'] == "FINANCIAL":
        return "financial_agent"
    elif state['agent_type'] == "RISK":
        return "risk_agent"

# 9. Build the LangGraph Workflow
workflow = StateGraph(AgentState)

# Add nodes
workflow.add_node("planner", planner_agent)
workflow.add_node("company_agent", company_agent)
workflow.add_node("financial_agent", financial_agent)
workflow.add_node("risk_agent", risk_agent)

# Add edges (Planner -> Workers)
workflow.set_entry_point("planner")
workflow.add_conditional_edges(
    "planner",
    route_query,
    {
        "company_agent": "company_agent",
        "financial_agent": "financial_agent",
        "risk_agent": "risk_agent"
    }
)

# Workers -> End
workflow.add_edge("company_agent", END)
workflow.add_edge("financial_agent", END)
workflow.add_edge("risk_agent", END)

# Compile the graph
app = workflow.compile()
print("✓ Adaptive Agentic Framework Built!")

# 10. Test the Adaptive Routing
print("\n" + "="*50)
print("TEST 1: Simple Registration Query")
print("="*50)
result1 = app.invoke({"query": "Where is TCS registered?"})
print(f"\nFinal Answer: {result1['response']}")

print("\n" + "="*50)
print("TEST 2: Financial Query")
print("="*50)
result2 = app.invoke({"query": "What is the total debt of HDFC?"})
print(f"\nFinal Answer: {result2['response']}")
```

```python
import os
import warnings
from langchain_groq import ChatGroq
from langchain_community.graphs import Neo4jGraph
from langchain_community.chains.graph_qa.cypher import GraphCypherQAChain
from langchain_core.prompts.prompt import PromptTemplate
from langgraph.graph import StateGraph, END
from typing import TypedDict, Literal
from neo4j import GraphDatabase

warnings.filterwarnings("ignore")

# 1. Credentials
os.environ["GROQ_API_KEY"] = "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"
NEO4J_URI = "neo4j+s://240b1b1e.databases.neo4j.io"
NEO4J_USERNAME = "240b1b1e"
NEO4J_PASSWORD = "T0-hXHZMe0Oib5frLfcxD9vrFNTywqVOr_hp4ExFu74"
NEO4J_DATABASE = "240b1b1e"

# 2. Connect to Neo4j and LLM
graph = Neo4jGraph(url=NEO4J_URI, username=NEO4J_USERNAME, password=NEO4J_PASSWORD, database=NEO4J_DATABASE)
driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USERNAME, NEO4J_PASSWORD))
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0)

# 3. Define the LangGraph State
class AgentState(TypedDict):
    query: str
    agent_type: str
    response: str

# 4. The Planner Agent (Same as before)
def planner_agent(state: AgentState):
    print("--- PLANNER AGENT ANALYZING QUERY ---")
    query = state['query']
    routing_prompt = f"""
    Analyze the following user query and classify it into one of three categories:
    1. "COMPANY" - If the question is about registration, location, state, industry, or company status.
    2. "FINANCIAL" - If the question is about revenue, debt, profit, margins, or financial ratios.
    3. "RISK" - If the question asks to assess, evaluate, or explain the risk of a company.

    Query: "{query}"
    Respond with ONLY ONE WORD: COMPANY, FINANCIAL, or RISK.
    """
    response = llm.invoke(routing_prompt)
    agent_type = response.content.strip().upper()
    if agent_type not in ["COMPANY", "FINANCIAL", "RISK"]:
        agent_type = "COMPANY"
    print(f"-> Decision: Route to {agent_type} Agent")
    return {"agent_type": agent_type}

# 5. Worker Agents for Company & Financial (Using GraphRAG)
CYPHER_GENERATION_TEMPLATE = """
You are an expert Neo4j Cypher developer.
Graph Schema:
Node labels: Company, Industry, State, FinancialRecord
Relationships: (Company)-[:OPERATES_IN]->(Industry), (Company)-[:REGISTERED_IN]->(State), (Company)-[:HAS_FINANCIAL_RECORD]->(FinancialRecord)
Company properties: name, cin, ticker, status. FinancialRecord properties: revenue, net_profit, total_debt, debt_to_equity, profit_margins, period.
CRITICAL: ALWAYS use `TOLOWER(c.name) CONTAINS TOLOWER('user_input')` in the WHERE clause. NEVER use exact matches.
User Question: {question}
Cypher Query (Return ONLY Cypher code):
"""
cypher_prompt = PromptTemplate(template=CYPHER_GENERATION_TEMPLATE, input_variables=["question"])
graph_qa = GraphCypherQAChain.from_llm(llm=llm, graph=graph, verbose=False, allow_dangerous_requests=True, cypher_prompt=cypher_prompt)

def company_agent(state: AgentState):
    print("--- COMPANY AGENT EXECUTING ---")
    response = graph_qa.invoke({"query": state['query']})
    return {"response": response['result']}

def financial_agent(state: AgentState):
    print("--- FINANCIAL AGENT EXECUTING ---")
    response = graph_qa.invoke({"query": state['query']})
    return {"response": response['result']}

# 6. UPGRADED RISK AGENT (Explainability + Confidence)
def risk_agent(state: AgentState):
    print("--- RISK AGENT EXECUTING ---")
    query = state['query']

    # Step A: Ask LLM to extract the company name from the user query
    extract_prompt = f"Extract only the company name from this query: '{query}'. Reply with ONLY the name, nothing else."
    company_name = llm.invoke(extract_prompt).content.strip()

    # Step B: Fetch raw data directly from Neo4j using a hardcoded query to ensure we get all fields for confidence
    cypher_query = """
    MATCH (c:Company)-[:HAS_FINANCIAL_RECORD]->(f:FinancialRecord)
    WHERE TOLOWER(c.name) CONTAINS TOLOWER($company_name)
    RETURN c.name AS name, c.status AS status, f.revenue AS revenue,
           f.net_profit AS profit, f.total_debt AS debt, f.debt_to_equity AS d2e
    LIMIT 1
    """
    with driver.session(database=NEO4J_DATABASE) as session:
        result = session.run(cypher_query, company_name=company_name)
        record = result.single()

    if not record:
        return {"response": f"Could not find data for {company_name} in the knowledge graph."}

    # Step C: Calculate Data Quality-Aware Confidence (Contribution #4)
    # Start at 100, subtract 15 for every missing critical field
    confidence = 100.0
    missing_fields = []
    if record['revenue'] is None or record['revenue'] == 0:
        confidence -= 15; missing_fields.append("Revenue")
    if record['profit'] is None:
        confidence -= 15; missing_fields.append("Net Profit")
    if record['debt'] is None or record['debt'] == 0:
        confidence -= 15; missing_fields.append("Total Debt")
    if record['d2e'] is None:
        confidence -= 15; missing_fields.append("Debt-to-Equity")

    # Step D: Evaluate Risk & Provide Evidence (Contribution #3)
    d2e = record['d2e'] if record['d2e'] else 0
    if d2e > 20:
        risk_level = "HIGH"
        evidence = f"Evidence: The Debt-to-Equity ratio is {d2e}, which is significantly above the safe benchmark of 20. This indicates high leverage."
    elif d2e > 5:
        risk_level = "MEDIUM"
        evidence = f"Evidence: The Debt-to-Equity ratio is {d2e}, which is moderate but requires monitoring."
    else:
        risk_level = "LOW"
        evidence = f"Evidence: The Debt-to-Equity ratio is {d2e}, which is within safe limits."

    # Check registration status as additional evidence
    status = record['status'] if record['status'] else "Unknown"
    if status != "ACTV":
        risk_level = "HIGH"
        evidence += f" Additional Evidence: Company registration status is '{status}', which is not active."

    # Step E: Format the final explainable output
    final_response = f"""
--- CORPORATE RISK ASSESSMENT ---
Company: {record['name']}
Risk Level: {risk_level}

{evidence}

Data Quality Confidence Score: {confidence}%
"""
    if missing_fields:
        final_response += f"\nNote: Confidence penalized due to missing data fields: {', '.join(missing_fields)}."

    return {"response": final_response}

# 7. Build the LangGraph Workflow
workflow = StateGraph(AgentState)
workflow.add_node("planner", planner_agent)
workflow.add_node("company_agent", company_agent)
workflow.add_node("financial_agent", financial_agent)
workflow.add_node("risk_agent", risk_agent)

workflow.set_entry_point("planner")
workflow.add_conditional_edges("planner", lambda state: state['agent_type'].lower() + "_agent", {
    "company_agent": "company_agent", "financial_agent": "financial_agent", "risk_agent": "risk_agent"
})
workflow.add_edge("company_agent", END)
workflow.add_edge("financial_agent", END)
workflow.add_edge("risk_agent", END)

app = workflow.compile()
print("✓ Explainable Agentic Framework Built!")

# 8. Test the New Risk Agent
print("\n" + "="*50)
print("TESTING EXPLAINABLE RISK AGENT")
print("="*50)
result = app.invoke({"query": "Assess the risk of Reliance"})
print(result['response'])
```

```python
import time
import os
import warnings
from langchain_groq import ChatGroq
from langchain_community.graphs import Neo4jGraph
from langchain_community.chains.graph_qa.cypher import GraphCypherQAChain
from langchain_core.prompts.prompt import PromptTemplate
from langgraph.graph import StateGraph, END
from typing import TypedDict, Literal
from neo4j import GraphDatabase

warnings.filterwarnings("ignore")

# --- [System Setup] ---
os.environ["GROQ_API_KEY"] = "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"
NEO4J_URI = "neo4j+s://240b1b1e.databases.neo4j.io"
NEO4J_USERNAME = "240b1b1e"
NEO4J_PASSWORD = "T0-hXHZMe0Oib5frLfcxD9vrFNTywqVOr_hp4ExFu74"
NEO4J_DATABASE = "240b1b1e"

graph = Neo4jGraph(url=NEO4J_URI, username=NEO4J_USERNAME, password=NEO4J_PASSWORD, database=NEO4J_DATABASE)
driver = GraphDatabase.driver(NEO4J_URI, auth=(NEO4J_USERNAME, NEO4J_PASSWORD))
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0)

CYPHER_GENERATION_TEMPLATE = """
You are an expert Neo4j Cypher developer.
Graph Schema: Node labels: Company, Industry, State, FinancialRecord. Relationships: (Company)-[:OPERATES_IN]->(Industry), (Company)-[:REGISTERED_IN]->(State), (Company)-[:HAS_FINANCIAL_RECORD]->(FinancialRecord).
Company properties: name, cin, ticker, status. FinancialRecord properties: revenue, net_profit, total_debt, debt_to_equity, profit_margins, period.
CRITICAL: ALWAYS use `TOLOWER(c.name) CONTAINS TOLOWER('user_input')` in the WHERE clause. NEVER use exact matches.
User Question: {question}
Cypher Query (Return ONLY Cypher code):
"""
cypher_prompt = PromptTemplate(template=CYPHER_GENERATION_TEMPLATE, input_variables=["question"])
graph_qa = GraphCypherQAChain.from_llm(llm=llm, graph=graph, verbose=False, allow_dangerous_requests=True, cypher_prompt=cypher_prompt)

# --- [Adaptive Pipeline] ---
class AgentState(TypedDict):
    query: str; agent_type: str; response: str; llm_calls: int

def planner_agent(state: AgentState):
    routing_prompt = f"Classify this query into COMPANY, FINANCIAL, or RISK: '{state['query']}'. Reply with ONLY ONE WORD."
    agent_type = llm.invoke(routing_prompt).content.strip().upper()
    if agent_type not in ["COMPANY", "FINANCIAL", "RISK"]: agent_type = "COMPANY"
    return {"agent_type": agent_type, "llm_calls": 1} # 1 call for planner

def company_agent(state: AgentState):
    res = graph_qa.invoke({"query": state['query']})
    return {"response": res['result'], "llm_calls": state['llm_calls'] + 2} # 1 for cypher, 1 for answer

def financial_agent(state: AgentState):
    res = graph_qa.invoke({"query": state['query']})
    return {"response": res['result'], "llm_calls": state['llm_calls'] + 2}

def risk_agent(state: AgentState):
    company_name = llm.invoke(f"Extract only the company name from this query: '{state['query']}'. Reply with ONLY the name.").content.strip()
    cypher_query = """
    MATCH (c:Company)-[:HAS_FINANCIAL_RECORD]->(f:FinancialRecord)
    WHERE TOLOWER(c.name) CONTAINS TOLOWER($company_name)
    RETURN c.name AS name, f.debt_to_equity AS d2e LIMIT 1
    """
    with driver.session(database=NEO4J_DATABASE) as session:
        record = session.run(cypher_query, company_name=company_name).single()
    d2e = record['d2e'] if record and record['d2e'] else 0
    risk_level = "HIGH" if d2e > 20 else ("MEDIUM" if d2e > 5 else "LOW")
    return {"response": f"Risk: {risk_level}", "llm_calls": state['llm_calls'] + 1} # 1 for extraction

workflow = StateGraph(AgentState)
workflow.add_node("planner", planner_agent)
workflow.add_node("company_agent", company_agent)
workflow.add_node("financial_agent", financial_agent)
workflow.add_node("risk_agent", risk_agent)
workflow.set_entry_point("planner")
workflow.add_conditional_edges("planner", lambda state: state['agent_type'].lower() + "_agent", {"company_agent": "company_agent", "financial_agent": "financial_agent", "risk_agent": "risk_agent"})
workflow.add_edge("company_agent", END); workflow.add_edge("financial_agent", END); workflow.add_edge("risk_agent", END)
adaptive_app = workflow.compile()

# --- [Fixed Pipeline Baseline] ---
def fixed_pipeline(query):
    calls = 0
    calls += 2; company_agent_res = graph_qa.invoke({"query": query})['result']
    calls += 2; financial_agent_res = graph_qa.invoke({"query": query})['result']
    calls += 1; risk_agent_res = risk_agent({"query": query, "llm_calls": 0})['response']
    return {"response": f"{company_agent_res} | {financial_agent_res} | {risk_agent_res}", "llm_calls": calls}

# --- [Run Experiment A] ---
test_questions = [
    "Where is TCS registered?",                     # Should be COMPANY
    "What is the total debt of HDFC?",               # Should be FINANCIAL
    "Assess the risk of Reliance",                   # Should be RISK
    "What is the revenue of Infosys?",               # Should be FINANCIAL
    "Give me the risk assessment of Bajaj Finance"  # Should be RISK
]

print("="*70)
print("EXPERIMENT A: ADAPTIVE vs FIXED PIPELINE")
print("="*70)
print(f"{'Query':<40} | {'Pipeline':<10} | {'Time (s)':<10} | {'LLM Calls':<10}")
print("-"*70)

for q in test_questions:
    # Run Adaptive
    start_time = time.time()
    adaptive_res = adaptive_app.invoke({"query": q, "llm_calls": 0})
    adaptive_time = time.time() - start_time
    adaptive_calls = adaptive_res['llm_calls']

    # Run Fixed
    start_time = time.time()
    fixed_res = fixed_pipeline(q)
    fixed_time = time.time() - start_time
    fixed_calls = fixed_res['llm_calls']

    # Print Results
    print(f"{q[:38]:<40} | {'Adaptive':<10} | {adaptive_time:<10.2f} | {adaptive_calls:<10}")
    print(f"{'':<40} | {'Fixed':<10} | {fixed_time:<10.2f} | {fixed_calls:<10}")
    print("-"*70)
```
